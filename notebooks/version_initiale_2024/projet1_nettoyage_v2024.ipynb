1. Importation des librairies utiles

2. Chargement du jeu de données "dtf_test"

```python
dtf=pd.read_csv("C:/Users/ismae/Desktop/Projets data Science/Dataset/dtf_test.csv")
# afficher les 05 premières lignes des données 
dtf.head(5)
```

```python
print("La taille de la BD  est: {}".format(dtf.shape))
```

3. Élimination des valeurs extrêmes (températures sèche ou humide de précisément +60°C ou -40°C, qui sont en réalité des erreurs de mesure du capteur).

```python
# éliminer les valeurs extremes de temerature 60 
masque1=(dtf['T2M_D_med']== 60) | (dtf["T2M_U_med"]==60)
dtf=dtf[~ masque1]
print("La nouvelle taille de la BD après avoir applique ce filtre est: {}".format(dtf.shape)) 
```

```python
# éliminer les valeurs extremes de temerature -40 
masque_1=(dtf['T2M_D_med']== -40) | (dtf["T2M_U_med"]==-40)
dtf=dtf[~ masque_1]
print("La nouvelle taille de la BD après avoir applique ce filtre est: {}".format(dtf.shape))
```

```python
# Vérification des valeurs suprimées 
dtf.describe()
```

4. Restriction à la période concernée par le gel (novembre - mai)

```python
print("la date la plus ancienne avant le filtre : {}".format(np.min(dtf['date'])))
```

```python
print("la date la plus recente avant le filtre {}".format(np.max(dtf['date'])))
```

```python
# Convertir en format datetime, on peut avoir certains formats en chaine de caractères 
dtf['date']=pd.to_datetime(dtf['date'])
```

```python
# Restriction de la periode concernée(novembre-mai), la periode de gel
dtf = dtf[(dtf['date'].dt.month >= 11) | (dtf['date'].dt.month <= 5)]

dtf.head(5)
```

```python
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

```python
print("la date la plus recente après le filtre {}".format(np.max(dtf['date'])))
```

5. Suppression des données diurnes (SSI > 50 W/m²) 

```python
masque2=dtf['SSI']>50
dtf=dtf[~masque2]
print("La nouvelle taille de la BD après avoir applique ce filtre est: {}".format(dtf.shape))
```

```python
# vérifications 
dtf.describe()
```

6. Suppression des données où la température humide est supérieure à la température sèche (T2M_U > T2M_D + 0.5 [marge prenant en compte l’erreur de mesure]) ;

```python
masque3=dtf["T2M_U_med"]>dtf["T2M_D_med"]+0.5
dtf=dtf[~ masque3]
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

7. Suppression des données éloignées des situations de gel (T2M > 10°C) ;
    

```python
masque4=dtf['T2M']>10
dtf=dtf[~ masque4]
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

```python
# verification 
dtf.describe()
```

8. Suppression des données où températures atmosphériques et températures sèches sont trop éloignées (|T2M - T2M_D - 1.5| > 3°C) ;

```python
masque5=np.abs(dtf["T2M"]-dtf["T2M_D_med"]-1.5)>3
dtf=dtf[~ masque5]
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

9. Suppression des données manquantes sur les variables T2M_U_med, T2M_U_min, T2M_U_max

```python
# supression des valeurs manquantes sur T2M_U_med
# identification des valeurs manquantes 
dtf.isnull().sum()
```

```python
# Comme c'est la seule colonne présentant les valeurs manquantes, il faudra suprimer les lignes avec la fonction "dropna()"
dtf.dropna(inplace=True)
dtf.head(5)
```

10. Suppression des jours où le capteur émet strictement moins de 50 fois (en théorie entre 96 et 240 mesures par jour) ;

```python
# Le nombre d'enregistrement étant par heure, puisqu'une journée à 24 heures, on peut avoir en moyenne en théorie ( entre 3.91 et 10 par heures) qu'il faut suprimer
masque6=(dtf['nbre_records']<4) 
dtf=dtf[~masque6]
#dtf.head(5)
```

```python
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

```python
np.mean(dtf['nbre_records'])
```

11. Suppression des données de capteurs en cours de fabrication, calibration et test (stations situées à moins de 0.05° d’un point basé à Nantes de latitude 47.266 et de longitude -1.561).


```python
masque7= (dtf['WD']<0.05) | (dtf['lat']==47.266) | (dtf['lon']==-1.561)
dtf=dtf[~masque7]
print("La nouvelle taille de la BD après avoir appliqué ce filtre est: {}".format(dtf.shape))
```

```python
dtf.describe()
```
